# Task 14: NumPy Matrix Operations

**Objective:** Understand basic matrix operations used in machine learning by creating two matrices and performing addition, subtraction, matrix multiplication, and transpose operations.

```python
import numpy as np

# Create two matrices
matrix_a = np.array([[1, 2, 3], 
                     [4, 5, 6]])

matrix_b = np.array([[7, 8], 
                     [9, 10], 
                     [11, 12]])

matrix_c = np.array([[1, 2], 
                     [3, 4]])

print("Matrix A:\n", matrix_a)
print("\nMatrix B:\n", matrix_b)
print("\nMatrix C (Square matrix for element-wise demo):\n", matrix_c)
```

## Addition and Subtraction
Addition and subtraction are performed element-wise. Matrices must have the exact same shape.

```python
# Create another matrix of the same shape as matrix_c for addition/subtraction
matrix_d = np.array([[5, 6], 
                     [7, 8]])

# Addition
add_result = matrix_c + matrix_d
print("Addition (C + D):\n", add_result)

# Subtraction
sub_result = matrix_c - matrix_d
print("\nSubtraction (C - D):\n", sub_result)
```

## Matrix Multiplication
Matrix multiplication computes the dot product of rows from the first matrix and columns from the second matrix. The number of columns in the first matrix must equal the number of rows in the second matrix.

```python
# Matrix Multiplication using np.dot() or the @ operator
# Matrix A is 2x3 and Matrix B is 3x2. The result will be a 2x2 matrix.

mul_result_dot = np.dot(matrix_a, matrix_b)
mul_result_at = matrix_a @ matrix_b

print("Matrix Multiplication (A @ B):\n", mul_result_at)
```

## Transpose Operations
Transposing a matrix flips it over its diagonal, switching its row and column indices.

```python
# Transpose using the .T attribute
transpose_a = matrix_a.T

print("Original Matrix A (2x3):\n", matrix_a)
print("\nTransposed Matrix A (3x2):\n", transpose_a)
```

## Interview Questions & Answers

# What is a matrix?   
A matrix is a two-dimensional rectangular array or grid of numbers, symbols, or expressions arranged in rows and columns. In programming and data science, it is commonly represented as a 2D array.

# What is the difference between element-wise and matrix multiplication?
Element-wise multiplication multiplies numbers at the exact same corresponding positions in two matrices of the same shape. Matrix multiplication (dot product) multiplies the rows of the first matrix by the columns of the second matrix, requiring the inner dimensions to match (e.g., a 2x3 matrix multiplied by a 3x2 matrix yields a 2x2 matrix).

## Why are matrices important in machine learning?
Matrices are the fundamental structure for representing datasets (rows as samples, columns as features). They allow for efficient, vectorized mathematical computations across massive amounts of data simultaneously, which is crucial for operations like updating weights in neural networks and performing linear algebra transformations in algorithms.
